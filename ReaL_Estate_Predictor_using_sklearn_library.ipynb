```python
import numpy as np
import pandas as pd
from sklearn.linear_model import SGDRegressor, LinearRegression
from sklearn.preprocessing import StandardScaler, PolynomialFeatures
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score
```

```python
df = pd.read_csv("/content/Real Estate.csv")
df
```

```python
X_train = np.array(df[['X2 house age','X3 distance to the nearest MRT station','X4 number of convenience stores','X5 latitude','X6 longitude']])
X_train
```

```python
y_train = np.array(df[['Y house price of unit area']])
y_train
```

```python
scaler = StandardScaler()
X_norm = scaler.fit_transform(X_train)
X_norm
```

```python
sgdr = SGDRegressor(max_iter=1000)
sgdr.fit(X_norm,y_train)
print(sgdr)
```

```python
b_norm = sgdr.intercept_
w_norm = sgdr.coef_
print(f"model parameters are: w={w_norm} b={b_norm}")
```

```python
m,_ = X_norm.shape
y_pred_sgd = sgdr.predict(X_norm)
for i in range(m):
  print(f"Prediction:{y_pred_sgd[i]:.1f}---Target Value:{y_train[i]}")
```

```python
score = r2_score(y_train,y_pred_sgd)
print("This model has",round(score,2)*100,"% accuracy")
```

BETTER APPROACH

```python
X_train,X_,y_train,y_ = train_test_split(X_train,y_train,test_size=0.40,random_state=1)
X_cv,X_test,y_cv,y_test = train_test_split(X_,y_,test_size=0.50,random_state=1)
del X_,y_
```

```python
print("X_train shape: ",X_train.shape)
print("y_train shape: ",y_train.shape)
print("X_cv shape: ",X_cv.shape)
print("y_cv shape: ",y_cv.shape)
print("X_test shape: ",X_test.shape)
print("y_test shape: ",y_test.shape)
```

```python
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
```

```python
model = LinearRegression()
model.fit(X_train_scaled,y_train)
```

```python
yhat = model.predict(X_train_scaled)
print("Traning MSE: ", mean_squared_error(y_train,yhat)/2)
```

```python
X_cv_scaled = scaler.transform(X_cv)
model.fit(X_cv_scaled,y_cv)
```

```python
yhat = model.predict(X_cv_scaled)
print("Validation MSE: ",mean_squared_error(y_cv,yhat)/2)
```

```python
poly = PolynomialFeatures(degree=2,include_bias=False)
X_train_mapped = poly.fit_transform(X_train)
```

```python
scaler_poly = StandardScaler()
X_train_mapped_scaled = scaler_poly.fit_transform(X_train_mapped)
```

```python
linear_model = LinearRegression()
linear_model.fit(X_train_mapped_scaled,y_train)
```

```python
yhat = linear_model.predict(X_train_mapped_scaled)
print("Traning MSE: ", mean_squared_error(y_train,yhat)/2)
```

```python
X_cv_mapped = poly.fit_transform(X_cv)
X_cv_mapped_scaled = scaler_poly.transform(X_cv_mapped)

linear_model.fit(X_cv_mapped_scaled,y_cv)

yhat = linear_model.predict(X_cv_mapped_scaled)
print("Validation MSE: ",mean_squared_error(y_cv,yhat)/2)
```

```python
X_test_mapped = poly.fit_transform(X_test)
X_test_mapped_scaled = scaler_poly.transform(X_test_mapped)

yhat = linear_model.predict(X_test_mapped_scaled)
print("Testing MSE: ",mean_squared_error(y_test,yhat)/2)
```

```python
train_mses = []
cv_mses = []
scalers = []
models = []
for degree in range (1,11):
  poly = PolynomialFeatures(degree, include_bias = False)
  X_train_mapped = poly.fit_transform(X_train)

  scaler_poly = StandardScaler()
  X_train_mapped_scaled = scaler_poly.fit_transform(X_train_mapped)
  scalers.append(scaler_poly)

  model = LinearRegression()
  model.fit(X_train_mapped_scaled, y_train)
  models.append(model)

  yhat = model.predict(X_train_mapped_scaled)
  train_mse = mean_squared_error(y_train,yhat) / 2
  train_mses.append(train_mse)

  X_cv_mapped = poly.fit_transform(X_cv)
  X_cv_mapped_scaled = scaler_poly.transform(X_cv_mapped)

  yhat = model.predict(X_cv_mapped_scaled)
  cv_mse = mean_squared_error(y_cv,yhat) / 2
  cv_mses.append(cv_mse)
```

```python
cv_mses
```

```python
degree = np.argmin(cv_mses)
print(degree)
```

```python
degree = 2
poly = PolynomialFeatures(degree, include_bias = False)
X_test_mapped = poly.fit_transform(X_test)

X_test_mapped_scaled = scalers[degree - 1].transform(X_test_mapped)

yhat = models[degree - 1].predict(X_test_mapped_scaled)
test_mse = mean_squared_error(y_test, yhat) / 2

print(f"Traning MSE: {train_mses[degree - 1]}")
print(f"Cross Validation MSE: {cv_mses[degree - 1]}")
print(f"Test MSE: {test_mse}")
print('\n')
print(f"Testing accuracy is: {round(r2_score(y_test,yhat),2)*100}%")
```
