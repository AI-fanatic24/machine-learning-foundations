```python
import pandas as pd
import numpy as np
from xgboost import XGBClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score,confusion_matrix
```

```python
df = pd.read_csv('/content/mushrooms.csv')
df
```

```python
one_hot_features = ['cap-shape','cap-surface','cap-color','bruises','odor','gill-attachment','gill-spacing','gill-size','gill-color','stalk-shape','stalk-root','stalk-surface-above-ring','stalk-surface-below-ring','stalk-color-above-ring','stalk-color-below-ring','veil-type','veil-color','ring-number','ring-type','spore-print-color','population','habitat']
```

```python
df = pd.get_dummies(data=df,prefix=one_hot_features,columns=one_hot_features)
```

```python
df
```

```python
df.loc[df["class"] == "p","class"] = 0      # 0 = poisonous
df.loc[df["class"] == "e","class"] = 1      # 1 = edible
```

```python
df
```

```python
features = [x for x in df.columns if x not in ['class']]
```

```python
X_train,X_,y_train,y_ = train_test_split(df[features],df['class'],train_size=0.60,random_state=1)
X_val,X_test,y_val,y_test = train_test_split(X_,y_,train_size=0.50,random_state=1)
del X_,y_
```

```python
X_train = np.array(X_train,dtype='float32')
X_val = np.array(X_val,dtype='float32')
y_train = np.array(y_train,dtype='float32').reshape(-1,1)
y_val = np.array(y_val,dtype='float32').reshape(-1,1)
X_test = np.array(X_test,dtype='float32')
y_test = np.array(y_test,dtype='float32').reshape(-1,1)
```

```python
print("X_train.shape",X_train.shape)
print("y_train.shape",y_train.shape)
print("X_val.shape",X_val.shape)
print("y_val.shape",y_val.shape)
print("X_test.shape",X_test.shape)
print("y_test.shape",y_test.shape)
```

```python
xgb_model = XGBClassifier(n_estimators=200,learning_rate=0.2,verbosity=1,random_state=1)
xgb_model.fit(X_train,y_train,eval_set=[(X_val,y_val)],early_stopping_rounds=5)
```

```python
xgb_model.best_iteration
```

```python
print("Train accuracy: ",accuracy_score(xgb_model.predict(X_train),y_train))
print("Validation accuracy: ",accuracy_score(xgb_model.predict(X_val),y_val))
print("Test accuracy: ",accuracy_score(xgb_model.predict(X_test),y_test))
```

```python
train_pred = xgb_model.predict(X_train)
val_pred = xgb_model.predict(X_val)
test_pred = xgb_model.predict(X_test)
```

```python
print("Train Confusion Matrix:\n",confusion_matrix(y_train,train_pred))
print("Validation Confusion Matrix:\n",confusion_matrix(y_val,val_pred))
print("Test Confusion Matrix:\n",confusion_matrix(y_test,test_pred))
```
